```python
# 가장 유명한 음식 설명
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
llm = ChatOllama(model='llama3.2',
                temperature=0)# 일관적(보수적)답변
prompt_template = PromptTemplate(
    template='{country}에서 가장 유명한 음식은 무엇입니까? 하나의 음식만 출력하세요',
    input_variables = ['country']
)
output_parser = StrOutputParser() # AIMessage()를 str 변환
output_parser.invoke(llm.invoke(prompt_template.invoke({'country':'Italy'})))
```

```python
# 음식 설명 랭체인
country_food_chain = prompt_template | llm | output_parser
country = input('입력할 나라는? :')
food_data = food_chain.invoke({'country':country})
food_data
```

```python
# 음식 이름만 받기
food_template = PromptTemplate(
    template = '{food}설명에서 설명하는 음식 이름 하나만 출력해 주세요',
    input_variables=['food']
)
food_parser = StrOutputParser()
food_parser.invoke(llm.invoke(food_template.invoke({'food' : food_data})))
```
