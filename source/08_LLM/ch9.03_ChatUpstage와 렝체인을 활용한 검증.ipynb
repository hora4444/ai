# 2. LLM 답변 생성
## 2.1 OpenAI SDK를 사용

```python
from openai import OpenAI # openai==1.52.2
import os
client = OpenAI(
    api_key=os.getenv("UPSTAGE_API_KEY"),
    base_url="https://api.upstage.ai/v1"
)
```

```python
response = client.chat.completions.create(
    model='solar-pro2',
    messages=[
        {'role':'system', 'content':'너는 친절하게 대답해주는 비서야'},
        {'role':'user', 'content':'2020 월드 시리즈 누가 우승했어?'}
    ],
    temperature=0.5,
    stream=False
)
```

```python
print(response.choices[0].message.content)
```

## 2.2 LangChain 선택
- 발급받은 API KEY를 "UPSTAGE_API_KEY" 라고 저장하면 별도의 설정없이 ChatUpstage를 바로 사용

```python
from langchain_upstage import ChatUpstage
from langchain_core.messages import SystemMessage, HumanMessage
llm = ChatUpstage()
messages = [
    SystemMessage(content = "너는 친절하게 대답해 주는 비서야"),
    HumanMessage(content="2020년 월드 시리즈 누가 우승했어?")
]
ai_message = llm.invoke(input=messages)
ai_message
```
